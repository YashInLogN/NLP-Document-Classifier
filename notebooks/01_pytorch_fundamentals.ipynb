```python
import torch
```

```python
# 0D
a = torch.tensor(5)

#1D
b= torch.tensor([1, 2])

#2D
c = torch.tensor([
    [1, 2],
    [3, 4]
])

print(a)
print(f"\n{b}")
print(f"\n{c}")
```

```python
x = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print("Number of dimension: ", x.ndim)
print("\nShape: ", x.shape)
print("\nDtype: ", x.dtype)
```

```python
documents = torch.randint(0, 1000, (100, 50))

print(documents.shape)
```

```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([10, 20, 30])

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

```python
result = torch.matmul(a, b)

print(result)
```
