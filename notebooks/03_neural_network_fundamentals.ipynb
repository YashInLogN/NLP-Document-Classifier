```python
import torch
import torch.nn as nn
```

```python
x = torch.tensor([5.0, 80.0, 70.0])

weights = torch.tensor([0.8, 0.4, 0.6])

bias = torch.tensor([-100.0])

z = torch.dot(x, weights) + bias

print("Neuron output: ", z)
```

```python
relu = nn.ReLU()

output = relu(z)

print(output)
```

```python
layer = nn.Linear(in_features=3, out_features=4)

x = torch.tensor([5.0, 80.0, 70.0])

output = layer(x)

print(output)

print("Output shape: ", output.shape)
```

```python
print("Weights: ", layer.weight)

print("\nBias: ", layer.bias)
```

```python
print("Weight shape: ", layer.weight.shape)
print("\nBias shape: ", layer.bias.shape)
```

```python
model = nn.Sequential(
    nn.Linear(3, 4),
    nn.ReLU(),
    nn.Linear(4, 1)
)

x = torch.tensor([[5.0, 80.0, 70.0]])

target = torch.tensor([[1.0]])

prediction = model(x)

loss_fn = nn.MSELoss()

loss = loss_fn(prediction, target)

print("Prediction: ", prediction)
print("\nLoss: ", loss)
```

```python
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

optimizer.zero_grad()

loss.backward()

optimizer.step()
```

```python
print(model[0].weight.grad)
```

```python
X_raw = torch.arange(1, 201, dtype=torch.float32).unsqueeze(1)

y_raw = 3 * X_raw + 2
X = X_raw/200
y = y_raw/602
print(X.shape)
print(y.shape)
```

```python
model = nn.Sequential(
    nn.Linear(1, 8),
    nn.ReLU(),
    nn.Linear(8, 1)
)
```

```python
loss_fn = nn.MSELoss()

optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
```

```python
epochs = 1500

for epoch in range(epochs):
    prediction = model(X)

    loss = loss_fn(prediction, y)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

    if (epoch+1) % 200 == 0:
        print(f"for epoch {epoch+1}: loss: {loss.item():.8f}")
```

```python
model.eval()
for v in [100.0, 200.0, 150.0, 50.0]:
    with torch.no_grad():
        test = torch.tensor([[v / 200]])
        result = model(test)

        original_pred = (result * 602)
    

    print(f"for val: {v}, Prediction: ", original_pred.item())
```

```python
X_raw = torch.arange(1, 1001, dtype=torch.float32).unsqueeze(1)
Z_raw = torch.arange(1, 1001, dtype=torch.float32).unsqueeze(1)
y_raw = (X_raw + Z_raw)/100

X = X_raw/1000
Z = Z_raw/1000
y = y_raw/20
```

```python
model = nn.Sequential(
    nn.Linear(2, 8),
    nn.ReLU(),
    nn.Linear(8, 1)
)

loss_fn = nn.MSELoss()

optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
```

```python
inputs = torch.cat((X, Z), dim=1)

epochs = 2000

for epoch in range(epochs):
    prediction = model(inputs)
    
    loss = loss_fn(prediction, y)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

    if (epoch+1) % 100 == 0:
        print(f"For epoch {epoch+1}: loss {loss.item():.8f}")
```

```python
model.eval()

with torch.no_grad():
    test = torch.tensor([[500.0/1000, 200.0/1000]])
    result = model(test)

    original_pred = result * 20

print(original_pred.item())
```
