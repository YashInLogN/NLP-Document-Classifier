```python
import torch
import torch.nn as nn
```

```python
x = torch.tensor([5.0, 80.0, 70.0])

weights = torch.tensor([0.8, 0.4, 0.6])

bias = torch.tensor([-100.0])

z = torch.dot(x, weights) + bias

print("Neuron output: ", z)
```

```python
relu = nn.ReLU()

output = relu(z)

print(output)
```

```python
layer = nn.Linear(in_features=3, out_features=4)

x = torch.tensor([5.0, 80.0, 70.0])

output = layer(x)

print(output)

print("Output shape: ", output.shape)
```

```python
print("Weights: ", layer.weight)

print("\nBias: ", layer.bias)
```

```python
print("Weight shape: ", layer.weight.shape)
print("\nBias shape: ", layer.bias.shape)
```

```python
model = nn.Sequential(
    nn.Linear(3, 4),
    nn.ReLU(),
    nn.Linear(4, 1)
)

x = torch.tensor([[5.0, 80.0, 70.0]])

target = torch.tensor([[1.0]])

prediction = model(x)

loss_fn = nn.MSELoss()

loss = loss_fn(prediction, target)

print("Prediction: ", prediction)
print("\nLoss: ", loss)
```

```python
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

optimizer.zero_grad()

loss.backward()

optimizer.step()
```

```python
X_raw = torch.arange(1, 201, dtype=torch.float32).unsqueeze(1)

y_raw = 3 * X_raw + 2
X = X_raw/200
y = y_raw/602
print(X.shape)
print(y.shape)
```

```python
model = nn.Sequential(
    nn.Linear(1, 8),
    nn.ReLU(),
    nn.Linear(8, 1)
)
```

```python
loss_fn = nn.MSELoss()

optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
```

```python
epochs = 1500

for epoch in range(epochs):
    prediction = model(X)

    loss = loss_fn(prediction, y)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

    if (epoch+1) % 200 == 0:
        print(f"for epoch {epoch+1}: loss: {loss.item():.8f}")
```

```python
model.eval()
for v in [100.0, 200.0, 150.0, 50.0]:
    with torch.no_grad():
        test = torch.tensor([[v / 200]])
        result = model(test)

        original_pred = (result * 602)
    

    print(f"for val: {v}, Prediction: ", original_pred.item())
```

```python
# Neural network for eqn: y = (x+y)/100
```

```python
torch.manual_seed(42)

X_values = torch.arange(1, 1001, dtype=torch.float32)
Z_values = torch.arange(1, 1001, dtype=torch.float32)

X_raw, Z_raw = torch.meshgrid(
    X_values,
    Z_values,
    indexing="ij"
)

X = X_raw.reshape(-1, 1)/ 1000
Z = Z_raw.reshape(-1, 1)/ 1000

y_raw = (X_raw+Z_raw)/100

y = y_raw.reshape(-1, 1)/20

inputs = torch.cat((X, Z), dim=1)

print("Inputs shape: ", inputs.shape)
print("y shape: ", y.shape)
```

```python
model = nn.Linear(2, 1)

# model = nn.Sequential(
#     nn.Linear(2, 8),
#     nn.ReLU(),
#     nn.Linear(8, 1)
# )

loss_fn = nn.MSELoss()

optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
```

```python
indices = torch.randperm(len(inputs))

train_size = int(0.7 * len(inputs))
val_size = int(0.15 * len(inputs))

train_idx = indices[:train_size]
val_idx = indices[train_size: train_size+ val_size]
test_idx = indices[train_size + val_size]

X_train = inputs[train_idx]
X_val = inputs[val_idx]
X_test = inputs[test_idx]

y_train = y[train_idx]
y_val = y[val_idx]
y_test = y[test_idx]

print(X_train.shape)
print(y_train.shape)
```

```python
epochs = 300

for epoch in range(epochs):
    prediction = model(X_train)

    loss = loss_fn(prediction, y_train)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    model.eval()

    with torch.no_grad():
        val_pred = model(X_val)

        val_loss = loss_fn(val_pred, y_val)

    if (epoch+1) % 50 == 0:
        print(f"for epoch {epoch+1}: Prediciton loss: {loss}, Validation loss: {val_loss}.")
```

```python
model.eval()

with torch.no_grad():
    test_prediction = model(X_test)

    test_loss = loss_fn(test_prediction, y_test)

print("Test Loss:", test_loss.item())

# Convert one prediction back to the original scale
test_input = torch.tensor([[0.5, 0.2]])

with torch.no_grad():
    prediction = model(test_input)

print("Predicted original value:", prediction.item() * 20)
```

```python
print("Weights: ", model.weight)
print("Bias: ", model.bias)
```

```python
X_raw = torch.arange(1, 101, dtype=torch.float32).unsqueeze(1)

X = X_raw/100

y_raw = 2 * X_raw ** 2 + 1

y = y_raw/20001
```

```python
model = nn.Sequential(
    nn.Linear(1, 16),
    nn.Tanh(),
    nn.Linear(16, 1)
)

loss_fn = nn.MSELoss()

optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
```

```python
print(X.shape)
print(y.shape)
```

```python
epochs = 5000

for epoch in range(epochs):

    prediction = model(X)

    loss = loss_fn(prediction, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch+1) % 50 == 0:
        print(f"For epoch {epoch+1}: Predicted loss {loss.item()}")
```

```python
model.eval()

test_input = torch.tensor([[0.4]])
with torch.no_grad():
    prediction = model(test_input)

print(f"Predicted original values: {prediction.item() * 20001}")
```
