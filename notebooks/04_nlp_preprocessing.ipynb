```python
import torch
import re
from torch.nn.utils.rnn import pad_sequence
```

```python
sentence = "Machine learning helps computer recognize patterns."

tokens = sentence.lower().replace(".", " ").split()

print(tokens)
```

```python
vocab = {
    word: idx
    for idx, word in enumerate(tokens)
}

tokens_id = [vocab[word] for word in tokens]

print(tokens_id)
```

```python
input_tensor = torch.tensor(tokens_id)

print("Shape: ", input_tensor.shape)
```

```python
def clean_text(text):

    text = text.lower().replace(".", " ")

    text = re.sub(r"[^\w\s]", "", text)

    text = " ".join(text.split())

    return text
```

```python
sentence = "Machine Learning is AMAZING!   Machine learning is transforming industries."

cleaned = clean_text(sentence)

print(cleaned)
```

```python
def tokenize_encode(text):

    tokens = text.split()

    vocab = {
        word: idx
        for idx, word in enumerate(tokens)
    }

    return [vocab[word] for word in tokens]

txt = "I'm learning NLP. It's fascinating!"
tokens_id = tokenize_encode(clean_text(txt))

tokens_id
```

```python
def chunk_sequence(sequence, chunk_size):
    return [
        sequence[i: i + chunk_size]
        for i in range(0, len(sequence), chunk_size)
    ]
```

```python
sequences = [i for i in range(10)]

padded = pad_sequence(
    [torch.tensor(sequences)],
    batch_first=True,
    padding_value=0
)

print(padded)
```

```python
chunks = chunk_sequence(sequences, 3)

padded = pad_sequence(
    [torch.tensor(chunk) for chunk in chunks],
    batch_first=True,
    padding_value = 0
)

print(padded)
```
