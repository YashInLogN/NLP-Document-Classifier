```python
import torch
```

```python
x = torch.tensor(3.0, requires_grad=True)

y = x ** 2

y.backward()

print(x.grad)
```

```python
k = torch.tensor(5.0, requires_grad=True)

a = k * 3

b = a + 5

c = b ** 2

c.backward()

print(k.grad)
```

```python
x = torch.tensor(5.0, requires_grad=True)

y = x * 5

print(y.grad_fn)
print(y.requires_grad)

y.backward()

print(x.grad)
```

```python
#  y = x**2 + z**2

x = torch.tensor(4.0, requires_grad=True)

z = torch.tensor(3.0, requires_grad=True)

y = x**2 + z**2

y.backward()

print("X gradient: ", x.grad)
print("Z gradient: ", z.grad)
```

```python
x = torch.tensor(5.0, requires_grad=True)

a = x * 2
b = a + 10
c = b ** 2

print(c.grad_fn)

c.backward()

print(x.grad)
```
